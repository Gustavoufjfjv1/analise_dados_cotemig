```python
import pandas as pd
import openpyxl
import matplotlib.pyplot  as plt
import seaborn as sns
```

```python
path = r"H:\Python\analise_dados\atividade 3\Tabela_3_Areas_gerais_de_formacao_na_graduacao.xlsx"
```

```python
bruto = pd.read_excel(path, header=None)
```

```python
bruto
```

```python
HomensMulheresRegiaoCursoSuperior = bruto.iloc[3:36, 0:4].copy()

HomensMulheresRegiaoCursoSuperior.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]

HomensMulheresRegiaoCursoSuperior = HomensMulheresRegiaoCursoSuperior.reset_index(drop=True)

# horas são número, não texto
for col in HomensMulheresRegiaoCursoSuperior.columns[1:]:
    HomensMulheresRegiaoCursoSuperior[col] = pd.to_numeric(HomensMulheresRegiaoCursoSuperior[col], errors="coerce")

HomensMulheresRegiaoCursoSuperior
```

```python
HomensMulheresRegiaoCursoSuperior['total_homens'] = HomensMulheresRegiaoCursoSuperior['homens'].sum()
```

```python
HomensMulheresRegiaoCursoSuperior['total_mulheres'] = HomensMulheresRegiaoCursoSuperior['mulheres'].sum()
```

```python
HomensMulheresRegiaoCursoSuperiorCTEM = bruto.iloc[3:36, [0, 4, 5, 6]].copy()

HomensMulheresRegiaoCursoSuperiorCTEM.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]

HomensMulheresRegiaoCursoSuperiorCTEM = HomensMulheresRegiaoCursoSuperiorCTEM.reset_index(drop=True)

for col in HomensMulheresRegiaoCursoSuperiorCTEM.columns[1:]:
    HomensMulheresRegiaoCursoSuperiorCTEM[col] = pd.to_numeric(HomensMulheresRegiaoCursoSuperiorCTEM[col], errors="coerce")

HomensMulheresRegiaoCursoSuperiorCTEM
```

```python
HomensMulheresRegiaoCursoSuperiorESSB = bruto.iloc[3:36, [0, 7, 8, 9]].copy()

HomensMulheresRegiaoCursoSuperiorESSB.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]

HomensMulheresRegiaoCursoSuperiorESSB = HomensMulheresRegiaoCursoSuperiorESSB.reset_index(drop=True)

for col in HomensMulheresRegiaoCursoSuperiorESSB.columns[1:]:
    HomensMulheresRegiaoCursoSuperiorESSB[col] = pd.to_numeric(HomensMulheresRegiaoCursoSuperiorESSB[col], errors="coerce")

HomensMulheresRegiaoCursoSuperiorESSB
```

```python
brasil = pd.DataFrame(
    {
        "area": ["Graduação (total)", "STEM", "Saúde / educação"],
        "homens": [
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "total_homens"].item(),
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "homens"].item(),
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "homens"].item(),
        ],
        "mulheres": [
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "total_mulheres"].item(),
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "mulheres"].item(),
            HomensMulheresRegiaoCursoSuperior.loc[HomensMulheresRegiaoCursoSuperior["regiao"] == "Brasil", "mulheres"].item(),
        ],
    }
)
```

```python
brasil_long = brasil.melt(id_vars="area", var_name="sexo", value_name="pessoas")
```

```python
fig, ax = plt.subplots(figsize=(9, 5))
sns.barplot(data=brasil_long, x="area", y="pessoas", hue="sexo", ax=ax)
ax.set_title("Brasil: pessoas com graduação, por área e sexo")
ax.set_xlabel("Área")
ax.set_ylabel("Pessoas")
plt.tight_layout()
plt.show()
```
