```python
HomensMulheresRegiaoCursoSuperior = bruto.iloc[3:36, 0:4].copy()

HomensMulheresRegiaoCursoSuperior.columns = [
    "regiao",
    "total",
    "homens",
    "mulheres",
]

HomensMulheresRegiaoCursoSuperior = HomensMulheresRegiaoCursoSuperior.reset_index(drop=True)

# horas são número, não texto
for col in HomensMulheresRegiaoCursoSuperior.columns[1:]:
    HomensMulheresRegiaoCursoSuperior[col] = pd.to_numeric(HomensMulheresRegiaoCursoSuperior[col], errors="coerce")

HomensMulheresRegiaoCursoSuperior
```
