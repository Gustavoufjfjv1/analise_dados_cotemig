```python
import pandas as pd
import openpyxl
import matplotlib.pyplot  as plt
import seaborn as sns
```

```python
path = r"H:\Python\analise_dados\atividade 4\CD2022_Populacao_2010_Compatibilizada_20231222.xlsx"
```

```python
bruto = pd.read_excel(path, header=2)
```

```python
bruto
```

```python
formatado = bruto.iloc[:5570, 1:]
formatado
```

```python
formatado.columns
```

```python
formatado["diferença"] = formatado["População Censo 2022"] - formatado["População Município 2010\n(Sinopse)"]
```

```python
porEstado = formatado.groupby("UF")[['População Município 2010\n(Sinopse)', 'População Censo 2022', "diferença"]].sum().reset_index()
```

```python
porEstado
```

```python
porEstado = porEstado.sort_values(by="diferença", ascending=False)
porEstado = porEstado.reset_index(drop=True)
porEstado.to_csv(r"H:\Python\analise_dados\atividade 4/formatadoPorCrescimentoEstado.xlsx", sep =";")
```

```python
porEstado
```

```python
porMunicipo = formatado.sort_values(by="diferença", ascending=False)
porMunicipo = porMunicipo.reset_index(drop=True)
porMunicipo.to_csv(r"H:\Python\analise_dados\atividade 4/formatadoPorCrescimentoMunicipio.xlsx", sep =";")
```

```python
porMunicipo
```

```python
fig, ax = plt.subplots(figsize=(9, 8))
ordem = porEstado.sort_values("População Censo 2022", ascending=False)
sns.barplot(data=ordem, y="UF", x="População Censo 2022", ax=ax, color="steelblue")
ax.set_title("População 2022 por UF")
ax.set_xlabel("Habitantes")
ax.set_ylabel("UF")
plt.tight_layout()
plt.show()
```
